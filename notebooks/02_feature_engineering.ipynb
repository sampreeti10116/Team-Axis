```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/raw_data.csv")

print(df.shape)
df.head()
```

```python
df['sowing_date'] = pd.to_datetime(df['sowing_date'])
```

```python
df['sowing_year'] = df['sowing_date'].dt.year
df['sowing_month'] = df['sowing_date'].dt.month
df['sowing_day'] = df['sowing_date'].dt.day
df['sowing_day_of_year'] = df['sowing_date'].dt.dayofyear
```

```python
df.drop('sowing_date', axis=1, inplace=True)
```

```python
df['temperature_range_c'] = (
    df['temperature_max_c'] -
    df['temperature_min_c']
)
```

```python
df['rainfall_per_day'] = (
    df['rainfall_mm'] /
    df['growing_season_days']
)
```

```python
df['total_npk_kg_ha'] = (
    df['nitrogen_kg_ha'] +
    df['phosphorus_kg_ha'] +
    df['potassium_kg_ha']
)
```

```python
df['np_ratio'] = (
    df['nitrogen_kg_ha'] /
    df['phosphorus_kg_ha'].replace(0, np.nan)
)
```

```python
df['kn_ratio'] = (
    df['potassium_kg_ha'] /
    df['nitrogen_kg_ha'].replace(0, np.nan)
)
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
df.drop('record_id', axis=1, inplace=True)
```

```python
print("Final Shape:", df.shape)
print(df.columns.tolist())
```
